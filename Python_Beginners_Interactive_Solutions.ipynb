## 1. Numbers, Variables, and Types

**Exercise 1:** Calculate the total cost from a price and quantity.

**Exercise 2:** Calculate an age five years from now.

```python
item_price = 12.50
quantity = 3
total_cost = item_price * quantity
print(f"Total cost: ${total_cost:.2f}")

age = 12
future_age = age + 5
print(f"In five years, you will be {future_age}.")
```

## 2. Text (Strings)

**Exercise 1:** Make a sentence lowercase and print its character count.

**Exercise 2:** Combine a first name and a favorite hobby into a sentence.

```python
sentence = "Python Is Fun"
lowercase_sentence = sentence.lower()
print(lowercase_sentence)
print(f"Characters: {len(lowercase_sentence)}")

first_name = "Alex"
hobby = "drawing"
print(f"{first_name}'s favorite hobby is {hobby}.")
```

## 3. Storing Many Values

**Exercise 1:** Add a fruit to a list and print the updated list.

**Exercise 2:** Add a `"course"` entry to a dictionary and print its value.

```python
snacks = ["apple", "banana"]
snacks.append("orange")
print(snacks)

learner = {"name": "Jordan"}
learner["course"] = "Python for Beginners"
print(learner["course"])
```

## 4. Making Decisions (if / else)

**Exercise 1:** Print whether a number is even or odd.

**Exercise 2:** Print `Pass` for a score of 60 or higher, and `Try again` otherwise.

```python
number = 7
if number % 2 == 0:
    print("Even")
else:
    print("Odd")

test_score = 72
if test_score >= 60:
    print("Pass")
else:
    print("Try again")
```

## 5. Repeating Things (Loops)

**Exercise 1:** Print the numbers 1 through 5 using `for` and `range`.

**Exercise 2:** Print a favorite-color sentence for each color in the list.

```python
for number in range(1, 6):
    print(number)

colors = ["blue", "green", "purple"]
for color in colors:
    print(f"My favorite color is {color}")
```

## 6. Functions

**Exercise 1:** Define `multiply` to return the product of two numbers.

**Exercise 2:** Convert Celsius to Fahrenheit with `C * 9 / 5 + 32`.

```python
def multiply(a, b):
    return a * b

print(multiply(4, 3))

def to_fahrenheit(celsius):
    return celsius * 9 / 5 + 32

print(to_fahrenheit(0))
```

## 7. Try It Yourself

**Challenge 1:** Write `double(number)` so it returns twice the number.

**Challenge 2:** Use a loop to print each name in the list.

```python
def double(number):
    return number * 2

print(double(5))
print(double(20))

names = ["Sam", "Priya", "Leo"]
for name in names:
    print(name)
```

## A few notes

- `*` multiplies numbers, and `+` adds numbers.
- `.lower()` returns a lowercase copy of a string; `len()` counts its characters.
- `.append()` adds to a list, while `dictionary[key] = value` adds or updates a dictionary entry.
- `%` gives a remainder, so `number % 2 == 0` checks whether a number is even.
- `range(1, 6)` stops before 6, so it produces 1 through 5.
- `return` sends a result back from a function.
