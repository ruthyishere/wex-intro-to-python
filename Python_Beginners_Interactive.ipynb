## 1. Numbers, Variables, and Types

Python can do math, store values in **variables**, and each value has a **type**.

```python
# Python can do math. Try running this.
print(10 + 5)    # add
print(10 - 5)    # subtract
print(10 * 5)    # multiply
print(10 / 5)    # divide
```

```python
# Variables store values so you can use them later.
name = "Alice"
age = 25

print(name)
print(age)
print(f"Next year, {name} will be {age + 1}")
```

```python
# Values have different types. Use type() to check.
print(type("Python"))   # text (str)
print(type(42))         # whole number (int)
print(type(3.14))       # decimal (float)
print(type(True))       # true/false (bool)
```

## 2. Text (Strings)

Text in Python is called a **string**. You can change it and combine it in useful ways.

```python
# Some handy things you can do with text
text = "hello world"

print(text.upper())                    # HELLO WORLD
print(text.replace("world", "Python")) # hello Python
print(len(text))                       # 11 (number of characters)
```

## 3. Storing Many Values

- **Lists** hold an ordered set of items: `["a", "b", "c"]`
- **Dictionaries** store labeled values (key → value): `{"name": "Bob"}`

```python
# Lists hold multiple items
fruits = ["apple", "banana", "orange"]

print(fruits)
print(fruits[0])        # first item (counting starts at 0)
print(len(fruits))      # how many items

fruits.append("grape")  # add a new item
print(fruits)
```

```python
# Dictionaries store labeled values (key: value)
student = {"name": "Bob", "age": 20}

print(student["name"])   # look up a value by its label
print(student["age"])

student["age"] = 21      # change a value
print(student)
```

## 4. Making Decisions (if / else)

Use `if`, `elif`, and `else` to run different code depending on a condition.

```python
# Change age and re-run to see different results
age = 18

if age >= 18:
    print("You are an adult")
else:
    print("You are not an adult yet")
```

```python
# More than two choices: use elif
score = 85

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
else:
    print("Grade: C or lower")
```

---

## 5. Repeating Things (Loops)

A **loop** repeats code so you don't have to type it over and over.

```python
# Repeat with a number range
for i in range(1, 4):   # 1, 2, 3
    print(i)
```

```python
# Loop through a list of items
fruits = ["apple", "banana", "orange"]

for fruit in fruits:
    print(f"I like {fruit}")
```

---

## 6. Functions

A **function** is reusable code you can run whenever you want. Define it with `def`.

```python
# Define a function once, then use it as many times as you like
def greet(name):
    print(f"Hello, {name}!")

greet("Alice")
greet("Bob")
```

```python
# Functions can give back a result with "return"
def add(a, b):
    return a + b

answer = add(5, 3)
print(answer)
```
